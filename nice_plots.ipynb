```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

from evo.core.trajectory import PoseTrajectory3D
from evo.core import lie_algebra, sync, metrics
from evo.core import trajectory
```

```python

# Define your trajectories
DPVO = np.loadtxt("/home/pellerito/DPVO_copy/Trajectories_to_plot/DPVO-abandonedfactory_night/Hard/P014/stamped_traj_estimate.txt")
DPVO_plus_e = np.loadtxt("/home/pellerito/DPVO_copy/Trajectories_to_plot/DPVO+events-abandonedfactory_night/Hard/P014/stamped_traj_estimate.txt")
RAMP = np.loadtxt("/home/pellerito/DPVO_copy/Trajectories_to_plot/RAMP-abandonedfactory_night/Hard/P014/stamped_traj_estimate.txt")
GT = np.loadtxt("/home/pellerito/DPVO_copy/Trajectories_to_plot/RAMP-abandonedfactory_night/Hard/P014/stamped_groundtruth.txt")

# Create a list of trajectory data
trajectories = [DPVO, DPVO_plus_e, RAMP, GT]

traj_gt = trajectory.PosePath3D(
            positions_xyz=GT[:, 1:4],
            orientations_quat_wxyz=GT[:, [7,4,5,6]],
        )

for trj in range(len(trajectories)-1):
    traj_est = trajectory.PosePath3D(
            positions_xyz=trajectories[trj][:, 1:4],
            orientations_quat_wxyz=trajectories[trj][:, [7,4,5,6]],
        )
    _ = traj_est.align(traj_gt, correct_scale=True)
    trajectories[trj] = traj_est.positions_xyz
trajectories[-1] = traj_gt.positions_xyz

# Define colors
colors = ['blue', 'orange', 'green', 'red']

# Create a figure and axis for the plot
fig, ax = plt.subplots()

# Set axis limits
ax.set_xlim(np.min([traj[:, 0].min() for traj in trajectories]),
            np.max([traj[:, 0].max() for traj in trajectories]))
ax.set_ylim(np.min([traj[:, 2].min() for traj in trajectories]),
            np.max([traj[:, 2].max() for traj in trajectories]))

# Initialize empty plots for each trajectory
lines = [ax.plot([], [], color=color, label=label, linewidth=1)[0] for color, label in zip(colors, ["DPVO", "DPVO+e", "RAMP", "GT"])]

# Initialize empty plots for each trajectory (dashed lines)
dashed_lines = [ax.plot([], [], color=color, linestyle='--', alpha=0.5, linewidth=1)[0] for color in colors]

# Initialize empty circles for each trajectory
circles = [ax.plot([], [], 'o', color=color, markersize=8)[0] for color in colors]

# Add legend
ax.legend(loc='lower left')

# Animation update function
def animate(frame):
    axis = 1 #2
    for i, line in enumerate(lines):
        x = trajectories[i][:frame+1, 0]
        y = trajectories[i][:frame+1, axis]
        line.set_data(x, y)
        circles[i].set_data(x[-1], y[-1])

        # Set the dashed line to show the full trajectory
        dashed_lines[i].set_data(
            trajectories[i][:, 0], 
            trajectories[i][:, axis])
        
    return lines + circles

# Create an animation
ani = animation.FuncAnimation(fig, animate, frames=min(len(traj) for traj in trajectories),
                              interval=100, blit=True)

# Specify high resolution settings
dpi = 600  # Adjust the DPI as needed for higher resolution

# Export the animation to an MP4 video in high resolution
ani.save('trajectory_animation_high_resolution.mp4', writer='ffmpeg', dpi=dpi)

# Show the plot (optional)
plt.show()

```
